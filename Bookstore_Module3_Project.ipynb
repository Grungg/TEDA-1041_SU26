```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = 'https://btech-data-analytics.github.io/bridgerland-technical-college/bookstore.html'
r = requests.get(url)
soup = BeautifulSoup(r.text, 'html.parser')
```

```python
data = []
for row in soup.find_all('tr', class_='book'):
    cols = row.find_all('td')
    data.append([cols[0].text.strip(), cols[1].text.strip(), cols[2].text.strip(), cols[3].text.strip(), int(cols[4].text.strip()), cols[5].text.strip(), float(cols[6].text.strip().replace('$', ''))])

df = pd.DataFrame(data, columns=['ISBN', 'Title', 'Author', 'Language', 'Pages', 'Topic', 'Price'])
df
```
