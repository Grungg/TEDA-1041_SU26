```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv(r'C:\Users\jcart\Downloads\CSV Files\companies.csv')
df.head()
```

```python
#Q1
plt.figure(figsize=(15,5))
sns.countplot(x='state', data=df)
plt.show()
```

```python
#Q2
df['state'].value_counts()
```

```python
#Q3
jobs = df.groupby('Industry')['job_openings'].sum()
plt.figure(figsize=(15,5))
plt.bar(jobs.index, jobs.values)
plt.xticks(rotation=90)
plt.show()
```

```python
#Q4
jobs.sort_values(ascending=False)
```

```python
#Q5
growth = df.groupby('founded')['growth_percentage'].mean()
plt.figure(figsize=(10,5))
plt.plot(growth.index, growth.values)
plt.show()
```
