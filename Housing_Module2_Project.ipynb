```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv(r'C:\Users\jcart\Downloads\CSV Files\housing.csv')
df.head()
```

```python
#Q1
sns.scatterplot(x='Landsize', y='Price', hue='Type', data=df)
plt.show()
```

```python
#Q2
df.groupby('Type')['Landsize'].corr(df['Price'])
```

```python
#Q3
year_price = df.groupby('YearBuilt')['Price'].mean()
plt.figure(figsize=(12,5))
plt.plot(year_price.index, year_price.values)
plt.show()
```

```python
#Q4
df['YearBuilt'].corr(df['Price'])
```

```python
#Q5
region_price = df.groupby('Regionname')['Price'].mean()
plt.figure(figsize=(12,5))
plt.bar(region_price.index, region_price.values)
plt.xticks(rotation=45)
plt.show()
```

```python
#Q6
region_price.sort_values(ascending=False)
```

```python
#Q7
plt.hist(df['Price'], bins=50)
plt.show()
```

```python
#Q8
df['Price'].skew()
```

```python
#Q9
plt.figure(figsize=(10,6))
sns.boxplot(x='YearBuilt', y='Regionname', data=df)
plt.show()
```
