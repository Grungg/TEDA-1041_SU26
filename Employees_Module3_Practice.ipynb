```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = 'https://btech-data-analytics.github.io/bridgerland-technical-college/employees_3.html'
r = requests.get(url)
soup = BeautifulSoup(r.text, 'html.parser')
```

```python
data = []
for row in soup.find_all('tr', class_='employee'):
    cols = row.find_all('td')
    data.append([cols[0].text.strip(), cols[1].text.strip(), int(cols[2].text.strip())])

df = pd.DataFrame(data, columns=['ID', 'Name', 'Age'])
df
```

```python
#Q1
df['Age'].mean()
```
